## Step 1: Load and Concatenate Data

```python

import numpy as np
import pandas as pd

Sales_2017 = pd.read_csv("Sales Transactions-2017.csv")
Sales_2018 = pd.read_csv("Sales Transactions-2018.csv")
Sales_2019 = pd.read_csv("Sales Transactions-2019.csv")

# Combine all datasets and reset the index cleanly
Total_sales_data = pd.concat([Sales_2017, Sales_2018, Sales_2019], ignore_index=True)
```

## Step 2: Drop Redundant Columns and Exact Duplicates

```python
# Drop Disc, Voucher Amount, and Gross columns
Total_sales_data.drop(["Disc", "Voucher Amount", "Gross"], axis=1, inplace=True)
Total_sales_data

# Drop duplicate rows
Total_sales_data.drop_duplicates(inplace=True)
```

```python
Total_sales_data
```

## Step 3: Clean Blank Spaces and Convert Empty Strings to NaN

```python
# 1. Clean extra spaces from text columns
for col in ["Date", "Voucher", "Party", "Product"]:
  Total_sales_data[col] = Total_sales_data[col].str.strip()

# 2. Convert empty strings to NaN
Total_sales_data.replace(r"^\s*$", np.nan, regex=True, inplace=True)
```

## Step 4: Drop Unwanted Rows (One line at a time)

```python
# 1. Drop rows that contain "Total" in the Voucher column
Total_sales_data = Total_sales_data[
    Total_sales_data["Voucher"].str.contains("Total", case=False, na=False)
    == False
]

# 2. Drop test rows that contain "dummy" in the Party name
Total_sales_data = Total_sales_data[
    Total_sales_data["Party"].str.contains("dummy", case=False, na=False)
    == False
]

# 3. Drop rows where Date or Voucher is missing (empty blank rows)[cite: 1]
Total_sales_data.dropna(subset=["Date", "Voucher"], inplace=True)
```

```python
Total_sales_data.info()
```

## Step 5: Clean and Convert Qty and Rate to Numbers

```python
# 1. Clean Qty: remove commas, then convert to numbers
Total_sales_data["Qty"] = (
    Total_sales_data["Qty"].astype(str).str.replace(",", "")
)
Total_sales_data["Qty"] = pd.to_numeric(
    Total_sales_data["Qty"], errors="coerce"
)

# 2. Clean Rate: remove commas, then convert to numbers
Total_sales_data["Rate"] = (
    Total_sales_data["Rate"].astype(str).str.replace(",", "")
)
Total_sales_data["Rate"] = pd.to_numeric(
    Total_sales_data["Rate"], errors="coerce"
)
```

```python
Total_sales_data.dtypes
```

## Step 6: Fix Date Format and Reset Index

```python
# 1. Convert Date text to actual datetime (dayfirst=True handles day/month/year format)
Total_sales_data["Date"] = pd.to_datetime(
    Total_sales_data["Date"], dayfirst=True, errors="coerce"
)

# 2. Reset the row numbers cleanly from 0 up to 95,223
Total_sales_data.reset_index(drop=True, inplace=True)
```

```python
Total_sales_data.info()
Total_sales_data.head()
```

## Step 7: Split the Voucher an make it as numbers

```python
# 1. Extract only digits from the Voucher column
Total_sales_data["Voucher"] = Total_sales_data["Voucher"].astype(str).str.extract(r'(\d+)')

# 2. Convert to integer type
Total_sales_data["Voucher"] = Total_sales_data["Voucher"].astype(int)
```

```python
# Extract digits safely
Total_sales_data["Voucher"] = Total_sales_data["Voucher"].astype(str).str.extract(r'(\d+)')

# Convert to integer safely (keeps integer format even if an entry is missing)
Total_sales_data["Voucher"] = pd.to_numeric(Total_sales_data["Voucher"], errors="coerce").astype("Int64")
```
